```python
import pandas as pd
import matplotlib.pyplot as plt

# a. Load and inspect the dataset
df = pd.read_csv(r"A:\Collge Code\Titanic-Dataset.csv")

# Display first 5 records
print(df.head())

# Display dataset information
print(df.info())

# Display number of rows and columns
print(df.shape)

# Check missing values
print(df.isnull().sum())


# b. Data cleaning

# Fill missing Age values with median
df['Age'] = df['Age'].fillna(df['Age'].median())

# Fill missing Embarked values with mode
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

# Drop Cabin column because it has many missing values
df.drop('Cabin', axis=1, inplace=True)

# Remove duplicate records
df.drop_duplicates(inplace=True)


# Data transformation

# Convert Sex into numerical values
df['Sex'] = df['Sex'].map({'male': 0, 'female': 1})

# Convert Embarked into numerical values
df['Embarked'] = df['Embarked'].map({
    'S': 0,
    'C': 1,
    'Q': 2
})

# Display data after preprocessing
print("\nAfter Preprocessing:")
print(df.head())


# c. Calculate summary statistics
for col in ['Age', 'Fare']:
    print("\nColumn:", col)
    print("Mean:", df[col].mean())
    print("Median:", df[col].median())
    print("Standard Deviation:", df[col].std())
    print("Variance:", df[col].var())


# d. Histogram of Age
plt.hist(df['Age'], bins=20, edgecolor='black')
plt.title("Age Histogram")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()


# Pie chart of survival
df['Survived'].value_counts().plot.pie(
    labels=['Not Survived', 'Survived'],
    autopct='%1.1f%%'
)

plt.title("Survival Distribution")
plt.ylabel("")
plt.show()
```
