```python
# Problem 8: Market Basket Analysis

import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules

# a. Load and preprocess data
df = pd.read_csv("Groceries_dataset.csv")
print(df.head())
transactions = df.groupby(
    ["Member_number", "Date"]
)["itemDescription"].apply(list)

# Convert transactions into one-hot format
items = sorted(set(item for t in transactions for item in t))

data = []
for t in transactions:
    row = {}
    for item in items:
        row[item] = item in t
    data.append(row)

basket = pd.DataFrame(data)

# b. Generate frequent itemsets using Apriori
frequent = apriori(
    basket,
    min_support=0.001,
    use_colnames=True
)

print("Frequent Itemsets:")
print(frequent.head(10))

# c. Generate association rules
rules = association_rules(
    frequent,
    metric="confidence",
    min_threshold=0.05
)

rules = rules[
    ["antecedents", "consequents",
     "support", "confidence", "lift"]
]

print("\nAssociation Rules:")
print(rules.sort_values("lift", ascending=False).head(10))

# d. Analysis
print("\nAnalysis:")
print("High support means the items occur frequently.")
print("High confidence means the consequent is likely when antecedent occurs.")
print("Lift greater than 1 indicates a positive association.")
```
