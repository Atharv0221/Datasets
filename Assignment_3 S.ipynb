```python
# Problem 3: t-Test and Chi-Square Test

import pandas as pd
from scipy.stats import ttest_ind, chi2_contingency

# Load dataset
df = pd.read_excel("A:\Collge Code\Student_Performance_Dataset.xlsx")

# 1. t-Test: Male vs Female Math Scores
male = df[df["Gender"] == "Male"]["Math_Score"]
female = df[df["Gender"] == "Female"]["Math_Score"]

t, p = ttest_ind(male, female, equal_var=False)

print("t-Test")
print("Male Mean:", round(male.mean(), 2))
print("Female Mean:", round(female.mean(), 2))
print("t-statistic:", round(t, 4))
print("p-value:", round(p, 4))

if p < 0.05:
    print("Significant difference")
else:
    print("No significant difference")

# 2. Chi-Square: Parental Education vs Performance
table = pd.crosstab(df["Parental_Education"], df["Performance_Level"])

chi, p, dof, expected = chi2_contingency(table)

print("\nChi-Square Test")
print(table)
print("Chi-square:", round(chi, 4))
print("p-value:", round(p, 4))

if p < 0.05:
    print("Significant association")
else:
    print("No significant association")
```
