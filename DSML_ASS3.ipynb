```python
# ==========================================================
# Practical: t-Test and Chi-Square Test
# Dataset: Student_Performance_Dataset.xlsx
# ==========================================================

import pandas as pd
from scipy.stats import ttest_ind, chi2_contingency
#Independent Two-Sample t-test:ttest_ind
#Chi-Square Test of Independence for a Contingency Table. It is used to determine
#whether there is a significant association (relationship) between two categorical variables.

# ----------------------------
# Load Dataset
# ----------------------------
df = pd.read_excel("A:\Collge Code\Student_Performance_Dataset.xlsx")

# ==========================================================
# PART 1 : Independent Samples t-Test
# Compare Math Scores of Male and Female Students
# ==========================================================

# Extract math scores
male_scores = df[df["Gender"] == "Male"]["Math_Score"]
female_scores = df[df["Gender"] == "Female"]["Math_Score"]

# Perform t-test
t_stat, p_value = ttest_ind(
    male_scores,
    female_scores,
    equal_var=False      # Welch's t-test
)
#performs an Independent
# Two-Sample t-test (Welch's t-test) to compare the average mathematics scores of male and female students.

print("=" * 60)
print("INDEPENDENT SAMPLE t-TEST")
print("=" * 60)

print("Mean Math Score (Male)   :", round(male_scores.mean(), 2))
print("Mean Math Score (Female) :", round(female_scores.mean(), 2))

print("\nT-statistic :", round(t_stat, 4))
print("P-value     :", round(p_value, 4))

alpha = 0.05

if p_value < alpha:
    print("\nResult: Reject the Null Hypothesis")
    print("There is a significant difference in mathematics scores.")
else:
    print("\nResult: Fail to Reject the Null Hypothesis")
    print("No significant difference in mathematics scores.")

    #H0:Null Hypothesis:No significant difference in mathematics scores
    #H1:significant difference in mathematics scores

# ==========================================================
# PART 2 : Chi-Square Test
# Relationship between Parental Education and Performance
# ==========================================================

print("\n" + "=" * 60)
print("CHI-SQUARE TEST OF INDEPENDENCE")
print("=" * 60)

# Create contingency table
# A contingency table counts how many observations belong to each combination of
#two categorical variables.

contingency_table = pd.crosstab(
    df["Parental_Education"],
    df["Performance_Level"]
)

print("\nContingency Table:\n")
print(contingency_table)

# Perform Chi-square test
chi2, p, dof, expected = chi2_contingency(contingency_table)

print("\nChi-Square Statistic :", round(chi2, 4))
print("Degrees of Freedom   :", dof)
print("P-value              :", round(p, 4))

if p < alpha:
    print("\nResult: Reject the Null Hypothesis")
    print("Parental education and student performance are significantly associated.")
else:
    print("\nResult: Fail to Reject the Null Hypothesis")
    print("No significant association exists between parental education and student performance.")

print("\nExpected Frequencies:\n")
print(pd.DataFrame(
    expected,
    index=contingency_table.index,
    columns=contingency_table.columns
))
```
